```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

# Extract variables
u = u_ds["u10"]
v = v_ds["v10"]

# Calculate wind speed
wind_speed = np.sqrt(u**2 + v**2)

print("========== WIND SPEED ==========")
print("Dimensions:", wind_speed.dims)
print("Shape:", wind_speed.shape)
print("Units:", wind_speed.attrs.get("units"))

print("Min:", wind_speed.min().item())
print("Max:", wind_speed.max().item())
print("Mean:", wind_speed.mean().item())
print("Median:", wind_speed.median().item())
print("Std:", wind_speed.std().item())
print("Missing:", wind_speed.isnull().sum().item())
```

```python
# ============================================================
# EXTREME WIND DAY
# ============================================================

daily_max_speed = wind_speed.max(
    dim=["latitude", "longitude"]
)

# Find integer index of maximum daily wind
max_day_index = daily_max_speed.argmax(dim="valid_time").item()

# Select the actual day using integer position
extreme_day = wind_speed.isel(
    valid_time=max_day_index
)

# Get the corresponding timestamp
max_day = wind_speed["valid_time"].isel(
    valid_time=max_day_index
).values

# Get maximum wind speed on that day
max_day_value = extreme_day.max().item()

print("========== MAXIMUM DAILY WIND ==========")
print("Index:", max_day_index)
print("Date:", max_day)
print("Maximum wind speed:", max_day_value, "m/s")
```

```python
# ============================================================
# FINAL WINDS QUALITY CONTROL
# ============================================================

print("================================================")
print("           FINAL WINDS QUALITY CONTROL")
print("================================================")

# ------------------------------------------------------------
# 1. Dimensions
# ------------------------------------------------------------

print("\n[1] Dimensions")
print("U:", u.dims, u.shape)
print("V:", v.dims, v.shape)
print("Wind speed:", wind_speed.dims, wind_speed.shape)


# ------------------------------------------------------------
# 2. Coordinate ranges
# ------------------------------------------------------------

print("\n[2] Coordinate ranges")

print(
    "Latitude:",
    float(u.latitude.min()),
    "to",
    float(u.latitude.max())
)

print(
    "Longitude:",
    float(u.longitude.min()),
    "to",
    float(u.longitude.max())
)

print(
    "Time:",
    u.valid_time.values[0],
    "to",
    u.valid_time.values[-1]
)


# ------------------------------------------------------------
# 3. Spatial resolution
# ------------------------------------------------------------

lat_diff = np.abs(np.diff(u.latitude.values))
lon_diff = np.abs(np.diff(u.longitude.values))

print("\n[3] Spatial resolution")

print("Unique latitude spacing:", np.unique(lat_diff))
print("Unique longitude spacing:", np.unique(lon_diff))


# ------------------------------------------------------------
# 4. Time spacing
# ------------------------------------------------------------

time_diff = np.diff(u.valid_time.values)

print("\n[4] Time spacing")

print(
    "Unique time intervals:",
    np.unique(time_diff / np.timedelta64(1, "D"))
)


# ------------------------------------------------------------
# 5. Missing values
# ------------------------------------------------------------

print("\n[5] Missing values")

print("U missing:", int(u.isnull().sum()))
print("V missing:", int(v.isnull().sum()))
print("Wind speed missing:", int(wind_speed.isnull().sum()))


# ------------------------------------------------------------
# 6. Infinite values
# ------------------------------------------------------------

print("\n[6] Infinite values")

print(
    "U infinite:",
    int(np.isinf(u.values).sum())
)

print(
    "V infinite:",
    int(np.isinf(v.values).sum())
)

print(
    "Wind speed infinite:",
    int(np.isinf(wind_speed.values).sum())
)


# ------------------------------------------------------------
# 7. Physical sanity check
# ------------------------------------------------------------

print("\n[7] Physical range")

print("U min/max:",
      float(u.min()),
      "/",
      float(u.max()))

print("V min/max:",
      float(v.min()),
      "/",
      float(v.max()))

print("Wind speed min/max:",
      float(wind_speed.min()),
      "/",
      float(wind_speed.max()))


# ------------------------------------------------------------
# 8. Duplicate coordinates
# ------------------------------------------------------------

print("\n[8] Duplicate coordinate check")

print(
    "Duplicate latitudes:",
    len(u.latitude) != len(np.unique(u.latitude.values))
)

print(
    "Duplicate longitudes:",
    len(u.longitude) != len(np.unique(u.longitude.values))
)

print(
    "Duplicate times:",
    len(u.valid_time) != len(np.unique(u.valid_time.values))
)


# ------------------------------------------------------------
# 9. U/V coordinate consistency
# ------------------------------------------------------------

print("\n[9] U/V consistency")

print(
    "Latitude identical:",
    np.array_equal(
        u.latitude.values,
        v.latitude.values
    )
)

print(
    "Longitude identical:",
    np.array_equal(
        u.longitude.values,
        v.longitude.values
    )
)

print(
    "Time identical:",
    np.array_equal(
        u.valid_time.values,
        v.valid_time.values
    )
)


print("\n================================================")
print("                 QC COMPLETE")
print("================================================")
```

```python
# ============================================================
# WINDS FINAL PROCESSED DATASET
# ============================================================

import os
import xarray as xr
import numpy as np

# ------------------------------------------------------------
# 1. Extract original variables
# ------------------------------------------------------------

u_wind = u.copy()
v_wind = v.copy()

# ------------------------------------------------------------
# 2. Calculate wind speed
# ------------------------------------------------------------

wind_speed = np.sqrt(u_wind**2 + v_wind**2)

# ------------------------------------------------------------
# 3. Add proper metadata
# ------------------------------------------------------------

u_wind.attrs = {
    "long_name": "10 metre U wind component",
    "standard_name": "eastward_wind",
    "units": "m s-1"
}

v_wind.attrs = {
    "long_name": "10 metre V wind component",
    "standard_name": "northward_wind",
    "units": "m s-1"
}

wind_speed.attrs = {
    "long_name": "Magnitude of daily mean 10 metre wind vector",
    "units": "m s-1"
}

# ------------------------------------------------------------
# 4. Create final dataset
# ------------------------------------------------------------

winds_final = xr.Dataset(
    {
        "u_wind": u_wind,
        "v_wind": v_wind,
        "wind_speed": wind_speed
    }
)

# ------------------------------------------------------------
# 5. Add dataset-level metadata
# ------------------------------------------------------------

winds_final.attrs = {
    "title": "Daily 10 metre Wind Components over the North Indian Ocean",
    "description": (
        "Daily mean 10 metre wind components and wind-speed "
        "magnitude over the North Indian Ocean."
    ),
    "spatial_domain": "5N-30N, 45E-105E",
    "spatial_resolution": "0.25 degree",
    "temporal_resolution": "Daily",
    "source": "ECMWF",
    "variables": "u_wind, v_wind, wind_speed",
    "wind_speed_definition": (
        "Magnitude of the daily mean U and V wind components"
    ),
    "project": "OceanF - Satellite Embedding-Based Deep Learning Framework"
}

# ------------------------------------------------------------
# 6. Create output directory
# ------------------------------------------------------------

output_dir = r"C:\OceanF\data\processed\Winds"

os.makedirs(output_dir, exist_ok=True)

output_file = os.path.join(
    output_dir,
    "Winds_processed.nc"
)

# ------------------------------------------------------------
# 7. Save NetCDF
# ------------------------------------------------------------

winds_final.to_netcdf(
    output_file,
    format="NETCDF4",
    encoding={
        "u_wind": {
            "dtype": "float32",
            "zlib": True,
            "complevel": 4
        },
        "v_wind": {
            "dtype": "float32",
            "zlib": True,
            "complevel": 4
        },
        "wind_speed": {
            "dtype": "float32",
            "zlib": True,
            "complevel": 4
        }
    }
)

print("================================================")
print("WINDS DATASET SAVED SUCCESSFULLY")
print("================================================")

print("File:")
print(output_file)

print("\nDataset:")
print(winds_final)
```
